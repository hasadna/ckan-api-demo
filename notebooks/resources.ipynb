```python
import os

import requests
from dotenv import load_dotenv

load_dotenv('../.env')

CKAN_URL = os.getenv('CKAN_URL')
CKAN_API_KEY = os.getenv('CKAN_API_KEY')
```

### Create Resource

```python
with open('../README.md', 'rb') as f:
    res = requests.post(
        CKAN_URL.rstrip('/') + '/api/3/action/resource_create',
        headers={
            'X-CKAN-API-Key': CKAN_API_KEY,
        },
        data={
            'package_id': 'api_demo',
            'name': 'קובץ דוגמה'
        },
        files={
            'upload': f
        }
    )
assert res.status_code == 200, f'{res.status_code} {res.text}'
```

### Update Resource

```python
res = requests.get(
    CKAN_URL.rstrip('/') + '/api/3/action/package_show',
    headers={
        'X-CKAN-API-Key': CKAN_API_KEY,
    },
    json={
        'id': 'api_demo'
    }
)
assert res.status_code == 200, f'{res.status_code} {res.text}'
result = res.json()['result']
resource_id = next((resource['id'] for resource in result['resources'] if resource['name'] == 'קובץ דוגמה'))
with open('../LICENSE', 'rb') as f:
    res = requests.post(
        CKAN_URL.rstrip('/') + '/api/3/action/resource_update',
        headers={
            'X-CKAN-API-Key': CKAN_API_KEY,
        },
        data={
            'id': resource_id,
            'package_id': 'api_demo',
            'name': 'קובץ דוגמה מעודכן',
        },
        files={
            'upload': f
        }
    )
```
